# CATTLEX: 01. Exploratory Data Analysis
This notebook conducts baseline statistical inspection of the cattle symptom disease reference dataset (2,044 samples, 93 symptom indicators, 26 target diseases).

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/Training.csv")
print("Dataset Shape:", df.shape)
print("Target diseases:", df["prognosis"].nunique())
```

```python
df["prognosis"].value_counts().plot(kind="bar", figsize=(12, 5), color="#3b82f6")
plt.title("Disease Frequency Distribution")
plt.tight_layout()
plt.show()
```
